```python
import numpy as np

# Coefficient matrix (A)
A = np.array([[2, 1],
              [3, 2]])

# Constant matrix (B)
B = np.array([10, 16])

# Calculate determinants
det_A = np.linalg.det(A)

# Determinants for x and y
A_x = A.copy()
A_x[:, 0] = B
det_Ax = np.linalg.det(A_x)

A_y = A.copy()
A_y[:, 1] = B
det_Ay = np.linalg.det(A_y)

# Cramer's Rule
x = det_Ax / det_A
y = det_Ay / det_A

print(f"Units of Product A (x): {x:.2f}")
print(f"Units of Product B (y): {y:.2f}")
```
