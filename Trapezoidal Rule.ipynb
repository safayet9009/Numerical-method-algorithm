```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Function to integrate
def f(x):
    return x**2

# Interval and number of subintervals
a, b = 0, 5   # integrate from 0 to 5
n = 10         # number of subintervals

# Step size
h = (b - a) / n

# x values
gx = np.linspace(a, b, n+1)
gy = f(gx)

# Trapezoidal Rule Formula
trap_integral = (h/2) * (gy[0] + 2*gy[1:-1].sum() + gy[-1])

# Put into pandas DataFrame
data = pd.DataFrame({"x": gx, "f(x)": gy})

print("Trapezoidal Rule Approximation of ∫x^2 dx from 0 to 5:", trap_integral)
print("\nData Table:")
print(data)

# Plot with trapezoids
plt.figure(figsize=(8,6))
plt.plot(gx, gy, 'b-', label='f(x) = x^2')
for i in range(n):
    xs = [gx[i], gx[i], gx[i+1], gx[i+1]]
    ys = [0, gy[i], gy[i+1], 0]
    plt.fill(xs, ys, 'skyblue', edgecolor='black', alpha=0.5)
plt.title("Trapezoidal Rule Approximation")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid()
plt.show()

# Bonus: Use scikit-learn regression to approximate the curve
X = gx.reshape(-1,1)
y = gy
model = LinearRegression()
model.fit(X, y)

print("Sklearn Linear Regression: f(x) ≈ %.2fx + %.2f" % (model.coef_[0], model.intercept_))
```
