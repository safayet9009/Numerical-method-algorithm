```python
import numpy as np; A = np.array([[3,2,-1,1],[2,-2,4,-2],[-1,0.5,-1,0]],float); n=len(A); 
for i in range(n): A[i]=A[i]/A[i,i]; 
for i in range(n): 
    for j in range(n): 
        if i!=j: A[j]=A[j]-A[j,i]*A[i]; 
print("Solutions:"); 
for i in range(n): print(f"x{i+1} = {A[i,-1]:.2f}")
```
